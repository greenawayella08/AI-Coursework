```python
from string import punctuation
import json
import re

#Load data source with articles
def load_articles(filename):
    with open(filename, "r", encoding="utf-8") as f:
        data = json.load(f)
    return data


#Sport rules used to make classification
sports_patterns = [
r"sport\w*",
r"game\w*",
r"opponent\w*",
r"cup",
r"goal\w*",
r"player\w*",
r"winner\w*",
r"race\w*",
r"team",
r"F1",
r"title\w*",
r"football\w*",
r"olympic\w*",
r"athlet\w*",
r"fan\w*",
r"federation\w*",
r"cycl\w*",
r"champion\w*",
r"sponso\w*",
r"cricke\w*",
]

#Politics rules used to make classification
politics_patterns = [
r"law\w*",
r"military\w*",
r"tech\w*",
r"intellig\w*",
r"Pentagon",
r"firm\w*",
r"labour\w*",
r"westminster",
r"election\w*",
r"conservative\w*",
r"minister\w*",
r"reform",
r"politic\w*",
r"parties",
r"debat\w*",
r"tax\w*",
r"parliament",
r"govern\w*",
]


def classify_article(text):
    import string

    print(text)

    text = text.translate(str.maketrans("","",string.punctuation))
    text = text.lower()

    sport_array = []
    politics_array = []
    sport_score = 0
    politics_score = 0

    #if there is a match then the score is increased
    for pattern in sports_patterns:
      if re.search(pattern, text):
        sport_array.append(pattern)
        sport_score += 1
    for pattern in politics_patterns:
      if re.search(pattern, text):
        politics_array.append(pattern)
        politics_score += 1

    #compares the scores and makes the relevant classification as well as outputs the words used to make classification
    if politics_score > sport_score:
        print("Reasons why it is classified as politics:")
        for i in range(len(politics_array)):
          print(f"\033[31m{politics_array[i]}\033[0m")
        print("Politics")
        return "Politics"
    elif sport_score > politics_score:
      print("Reasons why it is classified as sport:")
      for i in range(len(sport_array)):
        print(f"\033[34m{sport_array[i]}\033[0m")
      print("Sport")
      return "Sport"
    elif sport_score == politics_score and sport_score > 0:
        print("Reasons why it is classified as both:")
        for i in range(len(sport_array)):
          print(f"\033[34m{sport_array[i]}\033[0m")
        for i in range(len(politics_array)):
          print(f"\033[31m{politics_array[i]}\033[0m")
        print("Both")
        return "Both"
    else:
      print("Neither")
      return "Neither"


def evaluate(data):

    correct = 0
    total = 0
    #Splits the information in the file into article and label
    for item in data:

        text = item["article"]
        true_label = item["label"]

        predicted = classify_article(text)
        #determines whether the classification was correct
        if predicted == true_label:
            correct += 1

        total += 1
    #calculates the accuracy
    accuracy = correct / total

    print("Total articles:", total)
    print("Correct predictions:", correct)
    print("Accuracy:", round(accuracy, 3))


data = load_articles("SportvPolitics.json")

evaluate(data)
```
